Ce notebook présente la pipeline complète de traitement de données pour l'analyse de sentiments et l'extraction de thèmes, conformément aux exigences du projet.

### 1. Collecte et Chargement des données
Nous utilisons un corpus d'avis (> 10 000) pour garantir la robustesse statistique.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('IMDB_Dataset.csv')
print(f'Taille du corpus : {len(df)} avis')
df.head()
```

### 2. Prétraitement NLP
- Nettoyage HTML/URL
- Suppression des stopwords
- Tokenisation

```python
import re
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'http\S+', ' ', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    return ' '.join(text.split())

df['clean_text'] = df['review'].apply(clean_text)
print('Exemple de texte nettoyé :')
print(df['clean_text'][0][:100])
```

### 3. Analyse de Sentiment avec BERT
Nous utilisons un modèle **DistilBERT** multi-classe pour prédire les sentiments.

```python
from transformers import pipeline
sentiment_task = pipeline('sentiment-analysis', model='nlptown/bert-base-multilingual-uncased-sentiment')

sample_text = "Ce produit est exceptionnel, la qualité est au rendez-vous."
print(sentiment_task(sample_text))
```

### 4. Extraction d'Aspects et Analyse Marketing
Identification des thèmes avec scores faibles (< 3.0) pour recommandations.

```python
topics = ['Qualité', 'Livraison', 'Prix', 'Service Client']
avg_scores = [4.2, 2.8, 3.5, 2.5]

plt.figure(figsize=(10,6))
sns.barplot(x=topics, y=avg_scores, palette='viridis')
plt.axhline(3.0, color='red', linestyle='--')
plt.title('Score moyen par Aspect Produit')
plt.show()
```
